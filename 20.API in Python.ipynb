## 20. API in Python

##### What is an API?
`API(Application Programming Interface)` is a way for two application to communicate.

Example:

* Your Python apps send request to Google API
* Google sends using HTTP. Example: REST API

----

Types of APIs
--------

1. Web API (Most Common)
    `Used through internet using HTTP. Example: REST`

2. Library API
    `Example: Python math module function`

3. OS API
    `Example:Windows/Linux system calls`

4. Database API
    `Example:MySQL Connector, SQLite API`

### API Communication Model

##### API works on Request ↔ Response concept

Python sends:

* URL
* Method
* Data
* Header

Server returns:
* Status code
* Data(JSON/XML)
* Headers

----
### ➤ API Protocols

#### ⬩➤REST API (Most Used)
* Uses HTTP method
* Uses JSON format
### ⬩➤ SOAP API
* Uses XML
* Heavy and secure

### ⬩➤ GraphQL API
* Query based response
* Client chooses required filed

#### REST API (Most Used)

REST Means [Represntational State Transfer]

REST uses:
* URL [Uniform Resourse Locator]
* HTTP Method [Hyper Text Transfer Protocol]
* JSON Data 

Example 

`https://jsonplaceholder.typicode.com/users`

### HTTP Methods in REST API

-----
|Method|Purpose|Example|
|----|----|---|
|`GET`|Fetch data|Get all users|
|`POST`|Send/Create data|Add user|
|`PUT`|Update Full data|Replace all user information|
|`PATCH`|Update partical data|Update only emial |
|`DELETE`|Delete data|Delete user|

### POST API example (Create Data)

---


````python
url = "https://jsonplaceholder.typicode.com/posts"

data = {
    "title":"Python API Testing",
    "body":"Leaning POST Method",
    "userId":1   
}
response = requests.post(url,json=data)
print(response)

print(response.json())
````

### PUT API Example (Full Update)

````python
url = "https://jsonplaceholder.typicode.com/posts/1"

data = {
    "title":"Updated Title",
    "body":"Updated Body",
    "userId":1 
}

response= requests.put(url,json=data)
print(response)
print(response.json())
````

### PATCH API Example (Partial Update)

````python
url = "https://jsonplaceholder.typicode.com/posts/1"

data  = {
    'title':"Only Title Updated"
}
response = requests.patch(url,json=data)
print(response)
print(response.json())
````

````python
url = "https://jsonplaceholder.typicode.com/users"
# print(url)

try:
    response = requests.get(url)
    response.raise_for_status()
    print(response)
    users = response.json()
    # print(users)

    for user in users:
        print("User ID:",user['id'],"\n"
            "User Name:",user["name"],"\n",
              "User Email:",user['email'],"\n",
             "User City:",user["address"]["city"],"\n")
        
    
except requests.exceptions.RequestException as error:
    print("API Failed:",error)
    
````
