## 20. API in Python

##### What is an API?
`API(Application Programming Interface)` is a way for two application to communicate.

Example:

* Your Python apps send request to Google API
* Google sends using HTTP. Example: REST API

----

Types of APIs
--------

1. Web API (Most Common)
    `Used through internet using HTTP. Example: REST`

2. Library API
    `Example: Python math module function`

3. OS API
    `Example:Windows/Linux system calls`

4. Database API
    `Example:MySQL Connector, SQLite API`

### API Communication Model

##### API works on Request ↔ Response concept

Python sends:

* URL
* Method
* Data
* Header

* Data(JSON/XML)
* Headers

### HTTP Methods in REST API

-----
|Method|Purpose|Example|
|----|----|---|

|`PUT`|Update Full data|Replace all user information|
|`PATCH`|Update partical data|Update only emial |

### GET API Example in Python



````python
import requests
````

````python
print(requests.__version__)
````

````python
url = "https://jsonplaceholder.typicode.com/users"

response = requests.get(url)
print(response)
print(response.status_code)
print(response.json())
print(type(response.json()))
````

### POST API example (Create Data)

---


````python
url = "https://jsonplaceholder.typicode.com/posts"

data = {
    "title":"Python API Testing",
    "body":"Leaning POST Method",
    "userId":1   
}
response = requests.post(url,json=data)
print(response)

print(response.json())
````

### PUT API Example (Full Update)

````python
url = "https://jsonplaceholder.typicode.com/posts/1"

data = {
    "title":"Updated Title",
    "body":"Updated Body",
    "userId":1 
}

response= requests.put(url,json=data)
print(response)
print(response.json())
````

### PATCH API Example (Partial Update)

````python
url = "https://jsonplaceholder.typicode.com/posts/1"

data  = {
    'title':"Only Title Updated"
}
response = requests.patch(url,json=data)
print(response)
print(response.json())
````

#### DELETE API Example



````python
url = "https://jsonplaceholder.typicode.com/posts/1"

response = requests.delete(url)
print(response)
````
