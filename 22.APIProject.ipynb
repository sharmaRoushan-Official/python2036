```python
import requests
from datetime import datetime
from zoneinfo import ZoneInfo
from PIL import Image
def getWeather():
    api_keys = "f3d23ab84146eaf779dbe1fcb07bc122"
    cityName = input("Enter city name: ")
    url = f"https://api.openweathermap.org/data/2.5/weather?q={cityName}&appid=f3d23ab84146eaf779dbe1fcb07bc122"
    print(url)
    response = requests.get(url)
    print(response)
    if response.status_code == 200:
        data  = response.json()
        # print(data)
        print("=============Details Temperature===============")
        # print("==============Country Name===================")
        countryName = data["sys"]["country"]
        print("Country Name:", countryName)
        # print("=============City Name======================")
        cityName = data["name"]
        print("City Name:",cityName)
        # print("===============Minimum Temperature==========")
        temp = data['main']['temp']
        # print(temp)
        # Formula
        # 0K − 273.15 = -273.1°C
        tempCelsius = temp - 273.15
        print(f"Current Temperature:{tempCelsius:.2f}°C")
        # print("==================Feels Like=================")
        feelsLike = data["main"]["feels_like"]
        tempFeels = feelsLike-273.15
        print(f"Feels Like:{tempFeels:.2f}°C")
        # print(type(data))


        weather = data['weather'][0]['id']
        print(weather)


        # sunrise and sunset
       

        timestamp = data["sys"]["sunrise"]
        timestamp2 = data["sys"]["sunset"]
        
        ist_time = datetime.fromtimestamp(timestamp, ZoneInfo("Asia/Kolkata"))
        ist_time2 = datetime.fromtimestamp(timestamp2, ZoneInfo("Asia/Kolkata"))
        
        print("Sunrise:", ist_time.strftime("%d-%m-%Y %I:%M:%S %p"))
        print("Sunset:", ist_time2.strftime("%d-%m-%Y %I:%M:%S %p"))
        

        icon_code = data["weather"][0]["icon"]
        # Build the full image URL
        
    
        icon_url = f"https://openweathermap.org/img/wn/{icon_code}@2x.png"

        response = requests.get(icon_url)
        
        with open("weather_icon.png", "wb") as file:
            file.write(response.content)
        
        print("Image Saved")

getWeather()
```
